```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips_df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")
```

 1/ Display the 10 first rows of the dataset (no need to sort)

```python
tips_df.head(10)
```

```python
tips_df['day'].unique()
```

3/  What day of the week is there more bills? Plot this with a Seaborn Countplot.

```python
tips_df['day'].value_counts().head(1)
```

```python
order = tips_df['day'].value_counts().head(1)
sns.countplot( data=tips_df, x = 'day', order=order.index)
```

4/ Try to do some other countplots, varying x with one of the categorical column (sex, smoker, time

```python
order = tips_df['sex'].value_counts().head(1)
sns.countplot( data=tips_df, x = 'sex', order=order.index)
```

```python
order = tips_df['smoker'].value_counts().head(1)
sns.countplot( data=tips_df, x = 'smoker', order=order.index)
```

```python
order = tips_df['time'].value_counts().head(1)
sns.countplot( data=tips_df, x = 'time', order=order.index)
```

Let's use seaborn.FacetGrid

Run the cell below. What do you observe?
Change col in the first line with another column (e.g. "time"). Run the cell again. What do you observe?

```python
g = sns.FacetGrid(tips_df, col="day")
g.map(plt.hist, "total_bill")
```

```python
g = sns.FacetGrid(tips_df, col="time")
g.map(plt.hist, "total_bill")
```

Let's continue with FacetGrid and add a row="smoker" parameter. How many cells do you get in the plot?

```python
g = sns.FacetGrid(tips_df, col="time", row = 'smoker')
g.map(plt.hist, "total_bill")
```

Let's look at the data to see if our intuition is correct. We will do a scatterplot with x being total_bill and y the tip.

```python
with sns.axes_style(style="whitegrid"):
    sns.relplot(x="total_bill", y="tip", data=tips_df)
```

Another way of looking at this data is to use a seaborn.jointplot.

```python
with sns.axes_style("white"):
    sns.jointplot(x="total_bill", y="tip", kind="hex", data=tips_df)
```

A very useful tool to identify correlations is the seaborn.pairplot

```python
sns.pairplot(tips_df, height=2, hue="smoker")
```

We are not doing Machine Learning yet but we can use seaborn.lmplot to graphically read a linear correlation between two columns:

```python
sns.lmplot(x="total_bill", y="tip", col="smoker", data=tips_df)
```
